##### Flake and Black

**Note:** After running flake8 command you can see a lot of E501 errors. This behavior is caused by longer than allowed docstring lines, which black does not format (https://github.com/psf/black/issues/2289#issuecomment-1528473215, https://stackoverflow.com/questions/70966298/python-black-code-formatter-doesnt-format-docstring-line-length), and also we couldn't configure flake8 (both flake8 and flake8-pyproject) to not mark longer docstrings as an error. Thus on the second screenshot I simply used `--ignore=E501,E203,W503` (the last 2 added as `--ignore` flag overrides the existing config) to ignore this error. I consulted with my mentor and he said that this is an acceptable solution.
