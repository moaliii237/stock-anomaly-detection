```python
import pandas as pd
import numpy as np
from models.predict import ProductionAnomalyPredictor


def test_predictor_with_real_data():
    """Test the production predictor with your actual test data"""

    # Initialize the predictor
    print("Initializing Production Anomaly Predictor...")
    predictor = ProductionAnomalyPredictor(models_path="models/saved_models/")

    # Load your test data
    print("Loading test data...")
    df_test = pd.read_csv("test.csv")  # Your test file

    print(f"Loaded {len(df_test)} data points")
    print(f"Date range: {df_test['Date'].iloc[0]} to {df_test['Date'].iloc[-1]}")
    print(f"Features: {df_test.shape[1]} columns")

    # Convert DataFrame rows to dictionaries
    test_points = df_test.to_dict("records")

    print("\n" + "=" * 60)
    print("STARTING LIVE PREDICTION SIMULATION")
    print("=" * 60)

    results = []

    for i, data_point in enumerate(test_points):
        print(f"\nProcessing point {i+1}/{len(test_points)}: {data_point['Date']}")

        # Make prediction
        result = predictor.predict_single_point(data_point)
        results.append(result)

        # Display result
        if result["status"] == "buffering":
            print(f"{result['message']} - Progress: {result['progress']}")

        elif result["status"] == "success":
            print(f"Anomaly Risk: {result['anomaly_probability']:.1%}")

            if result["is_anomaly"]:
                print(
                    f"ANOMALY DETECTED: {result['predicted_anomaly_type']} "
                    f"(confidence: {result['confidence']:.1%})"
                )

                # Show type probabilities
                print("   Type probabilities:")
                for anomaly_type, prob in result["type_probabilities"].items():
                    print(f"   - {anomaly_type}: {prob:.1%}")
            else:
                print(
                    f"Normal market conditions (confidence: {result['confidence']:.1%})"
                )

        elif result["status"] == "error":
            print(f"Error: {result['message']}")

        # Add some spacing for readability
        if i > 0 and (i + 1) % 5 == 0:
            print("\n" + "-" * 40)

    # Summary analysis
    print("\n" + "=" * 60)
    print("PREDICTION SUMMARY")
    print("=" * 60)

    successful_predictions = [r for r in results if r["status"] == "success"]

    if successful_predictions:
        anomalies = [r for r in successful_predictions if r["is_anomaly"]]
        normal = [r for r in successful_predictions if not r["is_anomaly"]]

        print(f"Successful predictions: {len(successful_predictions)}")
        print(
            f"Anomalies detected: {len(anomalies)} ({len(anomalies)/len(successful_predictions)*100:.1f}%)"
        )
        print(
            f"Normal conditions: {len(normal)} ({len(normal)/len(successful_predictions)*100:.1f}%)"
        )

        if anomalies:
            print("\nDETECTED ANOMALIES:")
            for i, anom in enumerate(anomalies, 1):
                print(
                    f"   {i}. {anom['timestamp']} - {anom['predicted_anomaly_type']} "
                    f"({anom['anomaly_probability']:.1%})"
                )

        # Average confidence
        avg_anomaly_prob = np.mean(
            [r["anomaly_probability"] for r in successful_predictions]
        )
        print(f"\nAverage anomaly probability: {avg_anomaly_prob:.1%}")

        # Anomaly type distribution
        if anomalies:
            anomaly_types = [a["predicted_anomaly_type"] for a in anomalies]
            unique_types, counts = np.unique(anomaly_types, return_counts=True)
            print("\nAnomaly type distribution:")
            for anom_type, count in zip(unique_types, counts):
                print(f"   - {anom_type}: {count} times")

    else:
        print("No successful predictions made")

    return results, successful_predictions


def test_specific_sequence():
    """Test predictor with a specific sequence from your data"""

    print("\n" + "=" * 60)
    print("TESTING SPECIFIC SEQUENCE")
    print("=" * 60)

    # Initialize fresh predictor
    predictor = ProductionAnomalyPredictor(models_path="models/saved_models/")

    # Load test data
    df_test = pd.read_csv("test.csv")

    # Take first 30 points to build the sequence
    sequence_data = df_test.head(30).to_dict("records")

    print(
        f"Testing with sequence from {sequence_data[0]['Date']} to {sequence_data[-1]['Date']}"
    )

    result = None
    for i, point in enumerate(sequence_data):
        result = predictor.predict_single_point(point)

        if result["status"] == "buffering":
            print(f"Point {i+1}/30: Buffering... {result['progress']}")
        elif result["status"] == "success":
            print(f"\nFINAL PREDICTION for sequence ending at {point['Date']}:")
            print(f"   Anomaly probability: {result['anomaly_probability']:.1%}")
            print(f"   Is anomaly: {'YES' if result['is_anomaly'] else 'NO'}")
            if result["is_anomaly"]:
                print(f"   Predicted type: {result['predicted_anomaly_type']}")
                print(f"   Confidence: {result['confidence']:.1%}")
            break

    return result
```

```python
# Test with your actual data
print("TESTING PRODUCTION ANOMALY PREDICTOR")
print("Using your real test data from test.txt")

try:
    # Run full test
    results, successful = test_predictor_with_real_data()

    # Test specific sequence
    specific_result = test_specific_sequence()

    print("\nTesting completed successfully!")

except Exception as e:
    print(f"\nError during testing: {str(e)}")
```
